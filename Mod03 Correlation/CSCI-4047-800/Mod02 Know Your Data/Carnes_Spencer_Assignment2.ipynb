```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('./Income Data.csv')
print(df)
```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
min = np.min(df.income)
q1 = np.percentile(df.income, 25)
med = np.median(df.income)
mean = np.mean(df.income)
q3 = np.percentile(df.income, 75)
max = np.max(df.income)

print('------------ Income Data -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Mean: {mean}")
print(f"Q3: {q3}")
print(f"Max: {max}")

min = np.min(df.education)
q1 = np.percentile(df.education, 25)
med = np.median(df.education)
mean = np.mean(df.education)
q3 = np.percentile(df.education, 75)
max = np.max(df.education)

print('------------ Education Data -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Mean: {mean}")
print(f"Q3: {q3}")
print(f"Max: {max}")

min = np.min(df.prestige)
q1 = np.percentile(df.prestige, 25)
med = np.median(df.prestige)
mean = np.mean(df.prestige)
q3 = np.percentile(df.prestige, 75)
max = np.max(df.prestige)

print('------------ Prestige Data -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Mean: {mean}")
print(f"Q3: {q3}")
print(f"Max: {max}")
```

4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
plt.hist(df['prestige'], bins=20, color='orange', alpha=0.5)
plt.title("Prestige histogram")
plt.show()

plt.hist(df['education'], bins=20, color='orange', alpha=0.5)
plt.title("Education histogram")
plt.show()
```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
plt.scatter(x = df['income'], y = df['education'], alpha=0.3)
plt.title("Income compared to Education")
plt.xlabel('Income')
plt.ylabel("Education")
plt.show()

plt.scatter(x = df['education'], y = df['prestige'], alpha=0.3)
plt.title("Education compared to Prestige")
plt.xlabel('Education')
plt.ylabel("Prestige")
plt.show()
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
standard_dev = df['income'].std()
print('------------------Income data----------------------')
print(f"Standard Deviation: {standard_dev:.2f}")
variance = standard_dev ** 2 
print(f"Variance: {variance:.2f}")
```
